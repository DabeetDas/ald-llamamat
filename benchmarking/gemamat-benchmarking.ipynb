````python
def chunk_text(text, tokenizer, max_tokens=700, overlap=100):
    tokens = tokenizer.encode(text, add_special_tokens=False)

    chunks = []
    start = 0

    while start < len(tokens):
        end = min(start + max_tokens, len(tokens))
        chunk_tokens = tokens[start:end]

        chunks.append(
            tokenizer.decode(chunk_tokens, skip_special_tokens=True)
        )

        start += max_tokens - overlap

    return chunks
````

````python

# Kaggle-ready chunked extraction pipeline for LLaMAT-2 / LLaMA-2-style context limits.
# Assumes the earlier notebook cell has already loaded `tokenizer` and `model`.
# If not, run the model-loading cell above first.

from __future__ import annotations

import ast
import json
import math
import os
import random
import re
import traceback
import unicodedata
from collections import Counter, defaultdict
from copy import deepcopy
from pathlib import Path
from typing import Any

import torch
from tqdm.auto import tqdm

# -----------------------------
# Kaggle / runtime configuration
# -----------------------------

MODEL_MAX_INPUT_TOKENS = 2048       # LLaMA-2-style safe prompt length. Increase to 3072 if your GPU allows.
CHUNK_TOKENS = 700                  # Paper text tokens per chunk; prompt overhead needs room too.
CHUNK_OVERLAP_TOKENS = 120
MAX_NEW_TOKENS = 768
RANDOM_STATE = 42                  # Reproducible random paper selection from Data/.
LIST_MIN_VOTES = 1                  # 1 = recall-oriented union for list fields; set 2 for stricter voting.
MAX_LIST_ITEMS = 40

DEFAULT_KAGGLE_OUTPUT_DIR = Path('/kaggle/working/llamat2_chunked_extracted_info')
if not Path('/kaggle/working').exists():
    DEFAULT_KAGGLE_OUTPUT_DIR = Path('llamat2_chunked_extracted_info')

if 'tokenizer' not in globals() or 'model' not in globals():
    raise RuntimeError("Run the notebook model-loading cell first so `tokenizer` and `model` exist.")

if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token
if getattr(model, 'generation_config', None) is not None and model.generation_config.pad_token_id is None:
    model.generation_config.pad_token_id = tokenizer.pad_token_id
model.eval()

# -----------------------------
# Exact output schemas / defaults
# -----------------------------

AGENT_DEFAULTS: dict[str, dict[str, Any]] = {
    'summary': {
        'target_material': None,
        'process_type': None,
        'main_precursors': [],
        'temperature_range': None,
        'summary': None,
        'evidence': None,
    },
    'target_material': {
        'target_material': {
            'chemical_formula': None,
            'material_name': None,
            'material_class': None,
        },
        'evidence': None,
    },
    'precursor_coreactant': {
        'precursors': [],
        'coreactants': [],
        'purge_gas': [],
        'carrier_gas': [],
        'evidence': None,
    },
    'deposition_conditions': {
        'deposition_temperature_C': None,
        'pressure': '',
        'precursor_pulse_time_s': None,
        'coreactant_pulse_time_s': None,
        'purge_time_s': None,
        'number_of_cycles': None,
        'reactor_type': '',
        'evidence': None,
    },
    'reaction_conditions': {
        'reaction_equations': [],
        'surface_mechanism_description': '',
        'intermediate_species': [],
        'evidence': None,
    },
    'substrate_info': {
        'substrate_material': '',
        'substrate_orientation': '',
        'pretreatment': '',
        'surface_functionalization': '',
        'evidence': None,
    },
    'film_properties': {
        'film_thickness_nm': None,
        'density_g_cm3': None,
        'refractive_index': None,
        'surface_roughness_nm': None,
        'crystal_phase': '',
        'evidence': None,
    },
    'characterization': {
        'characterization_methods': [],
        'evidence': None,
    },
}

AGENT_NAMES = tuple(AGENT_DEFAULTS.keys())

SYSTEM_PREAMBLE = """\
You are a materials science expert specializing in Atomic Layer Deposition (ALD).

Your task is to extract structured scientific information from the provided research paper text.

Rules:
1. Only extract information explicitly supported by the text.
2. If information is missing, return null.
3. Do not infer or hallucinate values.
4. Include the exact evidence sentence from the paper when possible.
5. Return output strictly in JSON format.
6. Use standard scientific units where possible.
7. Do not return an all-empty placeholder template if the text explicitly contains the answer.
"""

AGENT_INSTRUCTIONS = {
    'summary': """
Task: Extract a concise summary of the Atomic Layer Deposition (ALD) process described in the text.

Focus on:
- material deposited
- main precursors
- process type (thermal ALD or PEALD)
- key temperature range
- purpose of the study

Output JSON schema:

{
  "target_material": "",
  "process_type": "",
  "main_precursors": [],
  "temperature_range": "",
  "summary": "",
  "evidence": ""
}
""",
    'target_material': """
Identify the primary material deposited using Atomic Layer Deposition. Ignore substrates, precursors, or reactor components. Focus ONLY on the primary target film being synthesized or studied.

Extract:
- chemical_formula: stoichiometric formula of the deposited material, e.g. "Al2O3", "TiO2". If not explicitly stated, infer only when the common name is chemically unambiguous.
- material_name: common English material name.
- material_class: broad class, e.g. "Oxide", "Nitride", "Pure Metal", "Sulfide", "2D Material".
- evidence: direct exact quote from the text.

Output JSON:

{
  "target_material": {
    "chemical_formula": "",
    "material_name": "",
    "material_class": ""
  },
  "evidence": ""
}
""",
    'precursor_coreactant': """
Extract the specific chemical inputs used during the Atomic Layer Deposition (ALD) process.

Definitions:
- precursors: primary metal, semiconductor, or organometallic sources, e.g. "TMA", "TiCl4", "Diethylzinc".
- coreactants: secondary reactants, e.g. "H2O", "O3", "NH3 plasma".
- purge_gas: inert gas used to evacuate the chamber between pulses.
- carrier_gas: inert gas used to transport precursor into the chamber; only list if explicitly described as carrier.

Rules:
1. If not explicitly mentioned, return an empty list [].
2. If multiple inputs are used, list all of them.
3. Provide one exact quote from the text as evidence.

Output JSON:

{
  "precursors": [],
  "coreactants": [],
  "purge_gas": [],
  "carrier_gas": [],
  "evidence": ""
}
""",
    'deposition_conditions': """
Extract the specific ALD process parameters from the text.

Rules:
1. Numerical Values: Extract only the number for fields ending in "_s" or "_C". If a range is provided, use a string like "250-300".
2. Timing Sequences: If sequence is like (0.1/5/0.1/5 s), map precursor pulse = 0.1, coreactant pulse = 0.1, purge = 5.
3. Reactor Type: Identify specific tool or reactor geometry.
4. If absent, return null for numerical fields and "" for text fields.
5. Evidence: direct, unedited quote containing these parameters.

Output JSON:

{
  "deposition_temperature_C": null,
  "pressure": "",
  "precursor_pulse_time_s": null,
  "coreactant_pulse_time_s": null,
  "purge_time_s": null,
  "number_of_cycles": null,
  "reactor_type": "",
  "evidence": ""
}
""",
    'reaction_conditions': """
Analyze the text to extract ALD surface reaction mechanism and chemistry.

Extract:
1. reaction_equations: formal chemical equations for surface reactions.
2. surface_mechanism_description: concise description of reaction steps.
3. intermediate_species: transient species formed during pulse/purge cycles.
4. evidence: exact excerpt where these reactions/mechanisms are discussed.

Rules:
- If no formal equations are provided, return [].
- Maintain original stoichiometry.

Output JSON:

{
  "reaction_equations": [],
  "surface_mechanism_description": "",
  "intermediate_species": [],
  "evidence": ""
}
""",
    'substrate_info': """
Extract information about the substrate used for the ALD process.

Extract:
1. substrate_material: base material, e.g. Silicon, Glass, Sapphire, PET, FTO-coated glass.
2. substrate_orientation: crystallographic orientation or Miller indices if provided.
3. pretreatment: cleaning/preparation before deposition.
4. surface_functionalization: chemical treatment to modify surface groups for nucleation.
5. evidence: exact quote describing substrate and preparation.

Output JSON:

{
  "substrate_material": "",
  "substrate_orientation": "",
  "pretreatment": "",
  "surface_functionalization": "",
  "evidence": ""
}
""",
    'film_properties': """
Extract physical and structural properties of the resulting ALD film.

Extract:
1. film_thickness_nm: measured thickness in nanometers. If multiple samples exist, prioritize primary sample or provide a range.
2. density_g_cm3: mass density in g/cm3.
3. refractive_index: refractive index n; include wavelength if mentioned.
4. surface_roughness_nm: RMS or Ra roughness in nm.
5. crystal_phase: crystallographic structure.
6. evidence: exact quote mentioning these results.

Output JSON:

{
  "film_thickness_nm": null,
  "density_g_cm3": null,
  "refractive_index": null,
  "surface_roughness_nm": null,
  "crystal_phase": "",
  "evidence": ""
}
""",
    'characterization': """
Identify characterization techniques used to analyze the ALD process or resulting films.

Rules:
1. characterization_methods: list all techniques explicitly used in the study. Include abbreviation and full name if provided.
2. Include structural, chemical, compositional, physical, optical, and electrical methods.
3. Ignore techniques only mentioned in cited works.
4. evidence: exact quote(s) listing or describing these tools.

Output JSON:

{
  "characterization_methods": [],
  "evidence": ""
}
""",
}

# -----------------------------
# JSON parsing and schema coercion
# -----------------------------

def clone_default(agent_name: str) -> dict[str, Any]:
    return json.loads(json.dumps(AGENT_DEFAULTS[agent_name]))


def looks_like_json_schema(value: Any) -> bool:
    if not isinstance(value, dict):
        return False
    schema_keys = {'type', 'properties', 'items', 'required', 'additionalProperties'}
    return bool(schema_keys.intersection(value.keys())) and (
        'properties' in value or value.get('type') in {'object', 'array'}
    )


def coerce_scalar(value: Any, template: Any) -> Any:
    if value is None:
        return None if template is None else deepcopy(template)
    if template is None:
        return value if isinstance(value, (str, int, float, bool)) else None
    if isinstance(template, str):
        return value if isinstance(value, str) else deepcopy(template)
    if isinstance(template, (int, float)):
        return value if isinstance(value, (int, float)) else deepcopy(template)
    if isinstance(template, bool):
        return value if isinstance(value, bool) else deepcopy(template)
    return value


def coerce_to_template(value: Any, template: Any) -> Any:
    if isinstance(template, dict):
        if isinstance(value, list):
            list_keys = [key for key, item in template.items() if isinstance(item, list)]
            if len(list_keys) == 1 and set(template.keys()) <= {list_keys[0], 'evidence'}:
                value = {list_keys[0]: value}
            else:
                raise ValueError('Model returned a list where an object was required.')
        if not isinstance(value, dict):
            raise ValueError('Model returned a non-object for a structured schema.')
        if not value:
            raise ValueError('Model returned an empty object.')
        if looks_like_json_schema(value):
            raise ValueError('Model returned a JSON Schema definition instead of extracted data.')
        overlapping = set(value.keys()).intersection(template.keys())
        if not overlapping:
            raise ValueError('Model output did not contain any expected schema keys.')
        return {
            key: coerce_to_template(value[key], template_value) if key in value else deepcopy(template_value)
            for key, template_value in template.items()
        }
    if isinstance(template, list):
        return value if isinstance(value, list) else deepcopy(template)
    return coerce_scalar(value, template)


def strip_model_response(text: Any) -> str:
    if hasattr(text, 'content'):
        text = text.content
    if not isinstance(text, str):
        return ''
    text = text.strip()
    text = re.sub(r'<s>|</s>|\[/?INST\]|<<SYS>>|<</SYS>>', '', text).strip()
    if '```json' in text:
        text = text.split('```json', 1)[1].split('```', 1)[0].strip()
    elif '```' in text:
        text = text.split('```', 1)[1].split('```', 1)[0].strip()
    return text


def extract_first_json_blob(text: str) -> str:
    text = strip_model_response(text)
    starts = [idx for idx in (text.find('{'), text.find('[')) if idx != -1]
    if not starts:
        raise ValueError(f'No JSON object or array found. Preview: {text[:300]!r}')
    start = min(starts)
    opener = text[start]
    closer = '}' if opener == '{' else ']'
    depth = 0
    in_string = False
    escape = False
    for idx in range(start, len(text)):
        ch = text[idx]
        if escape:
            escape = False
            continue
        if ch == '\\':
            escape = True
            continue
        if ch == '"':
            in_string = not in_string
            continue
        if in_string:
            continue
        if ch == opener:
            depth += 1
        elif ch == closer:
            depth -= 1
            if depth == 0:
                return text[start:idx + 1]
    # Fallback to greedy end if model omitted a final delimiter.
    return text[start:]


def robust_json_parse(text: Any, default: dict[str, Any]) -> dict[str, Any]:
    raw = strip_model_response(text)
    blob = extract_first_json_blob(raw)
    blob = re.sub(r',\s*([\]}])', r'\1', blob)
    blob = re.sub(r'\bNone\b', 'null', blob)
    blob = re.sub(r'\bTrue\b', 'true', blob)
    blob = re.sub(r'\bFalse\b', 'false', blob)
    parsed = None
    try:
        parsed = json.loads(blob)
    except json.JSONDecodeError:
        try:
            parsed = ast.literal_eval(blob)
        except Exception:
            parsed = None
    if parsed is None:
        raise ValueError(f'Could not parse model output as JSON. Preview: {raw[:300]!r}')
    return coerce_to_template(parsed, default)

# -----------------------------
# Text normalization / validation
# -----------------------------

def normalize_text(text: str) -> str:
    text = unicodedata.normalize('NFKC', text or '')
    text = text.replace('\u00a0', ' ')
    text = text.replace('\u2013', '-').replace('\u2014', '-').replace('\u2212', '-')
    text = re.sub(r'\s+', ' ', text)
    return text.strip().lower()


def compact_text(text: str) -> str:
    return re.sub(r'[^a-z0-9]+', '', normalize_text(text))


def evidence_is_in_text(fulltext: str, evidence: str) -> bool:
    if not isinstance(evidence, str) or not evidence.strip():
        return False
    normalized_evidence = normalize_text(evidence)
    normalized_fulltext = normalize_text(fulltext)
    if normalized_evidence in normalized_fulltext:
        return True
    compact_evidence = compact_text(evidence)
    compact_fulltext = compact_text(fulltext)
    return len(compact_evidence) >= 24 and compact_evidence in compact_fulltext


def has_meaningful_content(value: Any, ignore_keys: set[str] | None = None) -> bool:
    ignore_keys = ignore_keys or set()
    if isinstance(value, dict):
        return any(
            has_meaningful_content(child, ignore_keys=ignore_keys)
            for key, child in value.items()
            if key not in ignore_keys
        )
    if isinstance(value, list):
        return any(has_meaningful_content(item, ignore_keys=ignore_keys) for item in value)
    if isinstance(value, str):
        return bool(value.strip())
    return value is not None


def validate_paper_outputs(results: dict[str, Any], fulltext: str) -> dict[str, list[dict[str, Any]]]:
    issues: dict[str, list[dict[str, Any]]] = {}
    for agent_name, data in results.items():
        agent_issues = []
        if not isinstance(data, dict):
            agent_issues.append({'code': 'invalid_top_level_type', 'path': agent_name, 'message': f'Expected object, got {type(data).__name__}.'})
        else:
            evidence = data.get('evidence')
            extracted_content = {key: value for key, value in data.items() if key != 'evidence'}
            if has_meaningful_content(extracted_content):
                if not isinstance(evidence, str):
                    agent_issues.append({'code': 'missing_evidence', 'path': f'{agent_name}.evidence', 'message': 'Non-empty extraction is missing an evidence string.'})
                elif not evidence.strip():
                    agent_issues.append({'code': 'empty_evidence', 'path': f'{agent_name}.evidence', 'message': 'Non-empty extraction has an empty evidence field.'})
                elif not evidence_is_in_text(fulltext, evidence):
                    agent_issues.append({'code': 'evidence_not_found', 'path': f'{agent_name}.evidence', 'message': 'Evidence does not appear in source text.'})
        if agent_issues:
            issues[agent_name] = agent_issues
    return issues

# -----------------------------
# LLaMAT-2 / LLaMA-2 wrapper
# -----------------------------

def model_input_device(model_obj):
    try:
        return model_obj.get_input_embeddings().weight.device
    except Exception:
        return next(model_obj.parameters()).device


def normalize_tokenizer_output(tokenized) -> dict[str, torch.Tensor]:
    if hasattr(tokenized, 'keys'):
        return {key: tokenized[key] for key in tokenized.keys()}
    return {'input_ids': tokenized}


def build_plain_chat_prompt(messages: list[dict[str, str]]) -> str:
    chunks = []
    for message in messages:
        chunks.append(f"{message['role'].upper()}:\n{message['content']}")
    chunks.append('ASSISTANT:\n')
    return '\n\n'.join(chunks)


class KaggleLLaMAT2LLM:
    def __init__(self, model_obj, tokenizer_obj, max_input_tokens=MODEL_MAX_INPUT_TOKENS, max_new_tokens=MAX_NEW_TOKENS):
        self.model = model_obj
        self.tokenizer = tokenizer_obj
        self.max_input_tokens = max_input_tokens
        self.max_new_tokens = max_new_tokens
        self.raw_outputs: list[dict[str, Any]] = []

    def invoke(self, prompt: str, *, agent_name: str = 'unknown', chunk_index: int | None = None):
        messages = [
            {
                'role': 'system',
                'content': 'You extract Atomic Layer Deposition data. Return one valid JSON object only.',
            },
            {
                'role': 'user',
                'content': prompt + '\n\nReturn only valid JSON. Do not include markdown or explanation.',
            },
        ]
        if getattr(self.tokenizer, 'chat_template', None):
            try:
                tokenized = self.tokenizer.apply_chat_template(
                    messages,
                    add_generation_prompt=True,
                    return_tensors='pt',
                    truncation=True,
                    max_length=self.max_input_tokens,
                )
                inputs = normalize_tokenizer_output(tokenized)
            except Exception:
                prompt_text = build_plain_chat_prompt(messages)
                inputs = normalize_tokenizer_output(
                    self.tokenizer(prompt_text, return_tensors='pt', truncation=True, max_length=self.max_input_tokens)
                )
        else:
            prompt_text = build_plain_chat_prompt(messages)
            inputs = normalize_tokenizer_output(
                self.tokenizer(prompt_text, return_tensors='pt', truncation=True, max_length=self.max_input_tokens)
            )

        device = model_input_device(self.model)
        inputs = {key: value.to(device) for key, value in inputs.items()}
        with torch.inference_mode():
            generated = self.model.generate(
                **inputs,
                max_new_tokens=self.max_new_tokens,
                do_sample=False,
                pad_token_id=self.tokenizer.pad_token_id,
                eos_token_id=self.tokenizer.eos_token_id,
            )
        prompt_len = inputs['input_ids'].shape[-1]
        generated_tokens = generated[0][prompt_len:]
        raw = self.tokenizer.decode(generated_tokens, skip_special_tokens=True)
        cleaned = strip_model_response(raw)
        self.raw_outputs.append({'agent': agent_name, 'chunk_index': chunk_index, 'raw': raw, 'cleaned': cleaned})
        return type('AIMessage', (), {'content': cleaned})()


llm = KaggleLLaMAT2LLM(model, tokenizer)

# -----------------------------
# Chunking
# -----------------------------

def chunk_text_for_llama2(text: str, tokenizer_obj, max_tokens=CHUNK_TOKENS, overlap=CHUNK_OVERLAP_TOKENS) -> list[str]:
    token_ids = tokenizer_obj.encode(text, add_special_tokens=False)
    if not token_ids:
        return []
    chunks = []
    stride = max(1, max_tokens - overlap)
    for start in range(0, len(token_ids), stride):
        end = min(start + max_tokens, len(token_ids))
        chunks.append(tokenizer_obj.decode(token_ids[start:end], skip_special_tokens=True))
        if end >= len(token_ids):
            break
    return chunks

# -----------------------------
# Agent execution
# -----------------------------

def build_agent_prompt(agent_name: str, chunk: str) -> str:
    return SYSTEM_PREAMBLE + '\n' + AGENT_INSTRUCTIONS[agent_name] + '\nText:\n```\n' + chunk + '\n```'


def run_agent_on_chunk(agent_name: str, chunk: str, chunk_index: int, llm_obj: KaggleLLaMAT2LLM) -> tuple[dict[str, Any], str | None]:
    default = clone_default(agent_name)
    prompt = build_agent_prompt(agent_name, chunk)
    try:
        output = llm_obj.invoke(prompt, agent_name=agent_name, chunk_index=chunk_index)
        return robust_json_parse(output, default), None
    except Exception as exc:
        return default, f'{type(exc).__name__}: {exc}'

# -----------------------------
# Consensus / majority merge
# -----------------------------

def vote_key(value: Any) -> str:
    if isinstance(value, str):
        return normalize_text(value)
    return json.dumps(value, sort_keys=True, ensure_ascii=False).lower()


def is_empty_value(value: Any) -> bool:
    if value is None:
        return True
    if isinstance(value, str):
        return not value.strip()
    if isinstance(value, list):
        return len(value) == 0
    if isinstance(value, dict):
        return not has_meaningful_content(value)
    return False


def get_path(data: Any, path: tuple[str, ...]) -> Any:
    current = data
    for key in path:
        if not isinstance(current, dict) or key not in current:
            return None
        current = current[key]
    return current


def iter_template_leaf_paths(template: Any, prefix: tuple[str, ...] = ()) -> list[tuple[str, ...]]:
    if isinstance(template, dict):
        paths = []
        for key, child in template.items():
            if key == 'evidence':
                continue
            paths.extend(iter_template_leaf_paths(child, prefix + (key,)))
        return paths
    return [prefix]


def set_path(data: dict[str, Any], path: tuple[str, ...], value: Any) -> None:
    current = data
    for key in path[:-1]:
        current = current[key]
    current[path[-1]] = value


def flatten_non_evidence_values(data: Any) -> list[Any]:
    values = []
    if isinstance(data, dict):
        for key, child in data.items():
            if key == 'evidence':
                continue
            values.extend(flatten_non_evidence_values(child))
    elif isinstance(data, list):
        for item in data:
            values.extend(flatten_non_evidence_values(item))
    elif not is_empty_value(data):
        values.append(data)
    return values


def choose_scalar(candidates: list[tuple[Any, int, str | None]], fulltext: str) -> Any:
    if not candidates:
        return None
    grouped: dict[str, dict[str, Any]] = {}
    for value, chunk_index, evidence in candidates:
        key = vote_key(value)
        if key not in grouped:
            grouped[key] = {'value': value, 'count': 0, 'first_chunk': chunk_index, 'grounded': 0}
        grouped[key]['count'] += 1
        if isinstance(evidence, str) and evidence_is_in_text(fulltext, evidence):
            grouped[key]['grounded'] += 1
    winner = max(grouped.values(), key=lambda item: (item['count'], item['grounded'], -item['first_chunk']))
    return winner['value']


def choose_list(candidates: list[tuple[Any, int, str | None]], fulltext: str, min_votes=LIST_MIN_VOTES) -> list[Any]:
    grouped: dict[str, dict[str, Any]] = {}
    for value, chunk_index, evidence in candidates:
        if is_empty_value(value):
            continue
        key = vote_key(value)
        if key not in grouped:
            grouped[key] = {'value': value, 'count': 0, 'first_chunk': chunk_index, 'grounded': 0}
        grouped[key]['count'] += 1
        if isinstance(evidence, str) and evidence_is_in_text(fulltext, evidence):
            grouped[key]['grounded'] += 1
    selected = [item for item in grouped.values() if item['count'] >= min_votes]
    if not selected and grouped:
        # Keep the best candidate rather than dropping a true value that only appeared in one relevant chunk.
        selected = [max(grouped.values(), key=lambda item: (item['count'], item['grounded'], -item['first_chunk']))]
    selected.sort(key=lambda item: (-item['count'], -item['grounded'], item['first_chunk'], vote_key(item['value'])))
    return [item['value'] for item in selected[:MAX_LIST_ITEMS]]


def choose_evidence(chunk_outputs: list[dict[str, Any]], merged: dict[str, Any], fulltext: str) -> str | None:
    merged_keys = {vote_key(value) for value in flatten_non_evidence_values(merged)}
    candidates = []
    for idx, data in enumerate(chunk_outputs):
        if not isinstance(data, dict):
            continue
        evidence = data.get('evidence')
        if not isinstance(evidence, str) or not evidence.strip():
            continue
        data_keys = {vote_key(value) for value in flatten_non_evidence_values(data)}
        overlap = len(merged_keys.intersection(data_keys))
        grounded = evidence_is_in_text(fulltext, evidence)
        candidates.append((grounded, overlap, len(evidence), -idx, evidence))
    if not candidates:
        return None
    return max(candidates)[-1]


def merge_agent_outputs(agent_name: str, chunk_outputs: list[dict[str, Any]], fulltext: str) -> dict[str, Any]:
    template = clone_default(agent_name)
    merged = clone_default(agent_name)
    for path in iter_template_leaf_paths(template):
        template_value = get_path(template, path)
        values = []
        for chunk_index, data in enumerate(chunk_outputs):
            value = get_path(data, path)
            evidence = data.get('evidence') if isinstance(data, dict) else None
            if isinstance(template_value, list):
                if isinstance(value, list):
                    for item in value:
                        values.append((item, chunk_index, evidence))
            elif not is_empty_value(value):
                values.append((value, chunk_index, evidence))
        if isinstance(template_value, list):
            set_path(merged, path, choose_list(values, fulltext))
        else:
            chosen = choose_scalar(values, fulltext)
            if chosen is not None:
                set_path(merged, path, chosen)
    merged['evidence'] = choose_evidence(chunk_outputs, merged, fulltext)
    return coerce_to_template(merged, template)

# -----------------------------
# Paper and dataset processing
# -----------------------------

def read_fulltext(path: Path) -> str:
    return path.read_text(encoding='utf-8', errors='ignore')


def process_paper_chunked(
    paper_dir: Path,
    output_dir: Path = DEFAULT_KAGGLE_OUTPUT_DIR,
    *,
    llm_obj: KaggleLLaMAT2LLM = llm,
    chunk_tokens: int = CHUNK_TOKENS,
    overlap_tokens: int = CHUNK_OVERLAP_TOKENS,
) -> dict[str, Any]:
    txt_path = paper_dir / 'content.txt'
    if not txt_path.exists():
        return {'status': 'failed', 'paper': paper_dir.name, 'errors': {'paper': 'Missing content.txt'}}
    fulltext = read_fulltext(txt_path)
    if not fulltext.strip():
        return {'status': 'failed', 'paper': paper_dir.name, 'errors': {'paper': 'Empty content.txt'}}

    chunks = chunk_text_for_llama2(fulltext, tokenizer, max_tokens=chunk_tokens, overlap=overlap_tokens)
    paper_out = output_dir / paper_dir.name
    paper_out.mkdir(parents=True, exist_ok=True)
    chunk_out_dir = paper_out / '_chunk_agent_outputs'
    raw_out_dir = paper_out / '_raw_model_outputs'
    chunk_out_dir.mkdir(exist_ok=True)
    raw_out_dir.mkdir(exist_ok=True)

    results: dict[str, Any] = {}
    errors: dict[str, Any] = {}
    chunk_counts = {'chunks': len(chunks), 'chunk_tokens': chunk_tokens, 'overlap_tokens': overlap_tokens}

    for agent_name in tqdm(AGENT_NAMES, desc=f'{paper_dir.name}: agents', leave=False):
        parsed_chunks = []
        agent_errors = []
        start_raw_count = len(llm_obj.raw_outputs)
        for chunk_index, chunk in enumerate(tqdm(chunks, desc=f'{agent_name}: chunks', leave=False)):
            parsed, error = run_agent_on_chunk(agent_name, chunk, chunk_index, llm_obj)
            parsed_chunks.append(parsed)
            if error:
                agent_errors.append({'chunk_index': chunk_index, 'error': error})
        merged = merge_agent_outputs(agent_name, parsed_chunks, fulltext)
        results[agent_name] = merged
        if agent_errors:
            errors[agent_name] = agent_errors

        with (paper_out / f'{agent_name}.json').open('w', encoding='utf-8') as handle:
            json.dump(merged, handle, indent=2, ensure_ascii=False)
            handle.write('\n')
        with (chunk_out_dir / f'{agent_name}.json').open('w', encoding='utf-8') as handle:
            json.dump(parsed_chunks, handle, indent=2, ensure_ascii=False)
            handle.write('\n')
        for raw_record in llm_obj.raw_outputs[start_raw_count:]:
            if raw_record.get('agent') == agent_name:
                idx = raw_record.get('chunk_index')
                (raw_out_dir / f'{agent_name}_chunk_{idx}.txt').write_text(raw_record.get('raw', ''), encoding='utf-8')

    validation_issues = validate_paper_outputs(results, fulltext)
    validation_path = paper_out / 'validation_issues.json'
    if validation_issues:
        with validation_path.open('w', encoding='utf-8') as handle:
            json.dump(validation_issues, handle, indent=2, ensure_ascii=False)
            handle.write('\n')
    elif validation_path.exists():
        validation_path.unlink()

    errors_path = paper_out / 'agent_errors.json'
    if errors:
        with errors_path.open('w', encoding='utf-8') as handle:
            json.dump(errors, handle, indent=2, ensure_ascii=False)
            handle.write('\n')
    elif errors_path.exists():
        errors_path.unlink()

    with (paper_out / 'chunking_metadata.json').open('w', encoding='utf-8') as handle:
        json.dump(chunk_counts, handle, indent=2, ensure_ascii=False)
        handle.write('\n')

    issue_count = sum(len(items) for items in validation_issues.values())
    status = 'completed' if not errors else 'warning'
    if issue_count:
        status = 'warning'
    return {'status': status, 'paper': paper_dir.name, 'chunks': len(chunks), 'errors': errors, 'validation_issues': validation_issues}


def validate_data_dir(base_dir: Path) -> Path:
    base_dir = Path(base_dir).expanduser().resolve()
    if base_dir.name != 'Data':
        raise ValueError(f'This pipeline is restricted to a folder named Data/. Got: {base_dir}')
    if not base_dir.exists() or not base_dir.is_dir():
        raise FileNotFoundError(f'Data directory does not exist: {base_dir}')
    return base_dir


def resolve_paper_dirs(
    base_dir: Path,
    folders: list[str] | None = None,
    start: int = 0,
    stop: int | None = None,
    limit: int | None = None,
    random_state: int = RANDOM_STATE,
) -> list[Path]:
    base_dir = validate_data_dir(base_dir)
    if folders:
        if any(Path(folder).is_absolute() or '/' in folder or '\\' in folder for folder in folders):
            raise ValueError('folders must contain names from Data/ only, not paths.')
        dirs = [base_dir / folder for folder in folders]
    else:
        dirs = sorted(path for path in base_dir.iterdir() if path.is_dir())
        dirs = dirs[start:stop]
    random.Random(random_state).shuffle(dirs)
    if limit is not None:
        dirs = dirs[:limit]
    return dirs


def process_dataset_chunked(
    base_dir: Path,
    output_dir: Path = DEFAULT_KAGGLE_OUTPUT_DIR,
    *,
    folders: list[str] | None = None,
    start: int = 0,
    stop: int | None = None,
    limit: int | None = None,
    random_state: int = RANDOM_STATE,
) -> list[dict[str, Any]]:
    base_dir = validate_data_dir(base_dir)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    paper_dirs = resolve_paper_dirs(
        base_dir,
        folders=folders,
        start=start,
        stop=stop,
        limit=limit,
        random_state=random_state,
    )
    print(f'Random paper selection seed: {random_state}')
    run_log = []
    for paper_dir in tqdm(paper_dirs, desc='papers'):
        result = process_paper_chunked(paper_dir, output_dir=output_dir)
        run_log.append(result)
        with (output_dir / 'run_log.json').open('w', encoding='utf-8') as handle:
            json.dump(run_log, handle, indent=2, ensure_ascii=False)
            handle.write('\n')
    return run_log


def find_kaggle_content_roots(search_root: Path = Path('/kaggle/input')) -> list[Path]:
    if not search_root.exists():
        return []
    roots = set()
    for content_path in search_root.rglob('content.txt'):
        roots.add(content_path.parent.parent)
    return sorted(roots)

print('Chunked LLaMAT-2 extraction pipeline ready.')
print(f'Default output directory: {DEFAULT_KAGGLE_OUTPUT_DIR}')
print('If needed, candidate Kaggle data roots:', find_kaggle_content_roots()[:10])
````
